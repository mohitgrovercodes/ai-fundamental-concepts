## Understanding Exceptions

Exeption handling in Python allows one to handle errors gracefully and take corrective actions without stopping the execution of the program or crashing the code. This lesson covers the basics of exceptions, including how to use - try, except, else and finally blocks (clauses).

### What are exceptions?

Exceptions are events that dissrupt the flow of a program. They occur when an error is encountered during program execution. Common errors include:

* ZeroDivisionError: Dividing by zero is undefined
* FileNotFoundError: File not found or not exists.
* ValueError: When function recieves an argument of the correct type but an inappropriate (invalid) value.
* NameError: Local or global name is not found.
* IndexError: Sequence index is out of range.
* KeyError: Dictionary key is not found.


```python
## NameError
a = b
```

```python
## NameError Solution

try:
    a = b
except NameError as ex:
    print("Variable 'b' is not defined. Please define it before using it. " + str(ex))
```

```python
## ZeroDivisionError

c = 5 / 0
```

```python
## ZeroDivisionError Handling

try:
    c = 5 / 0
except ZeroDivisionError as ex:
    print("Division by zero is not allowed. Please provide a non-zero denominator. " + str(ex))
```

```python
## Handling Multiple Exceptions
## Using the Exception Class of Exception Handling which is the parent class. All exceptions are derived from it.
## Exception class should be used at the end after all specific exceptions are handled.

try:
    c = 5 / 3
    a = b
except ZeroDivisionError as ex:
    print(ex)
    print("Division by zero is not allowed. Please provide a non-zero denominator.")
except Exception as ex1:
    print(ex1)
    print("Main exception got caought here.")
```

```python
## Another example of handling multiple exceptions
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ValueError:
    print("Invalid input. Please enter a valid integer.")
except ZeroDivisionError:
    print("Division by zero is not allowed.")
except Exception as e:
    print(e)
```

```python
## try, except, else block
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ValueError:
    print("Invalid input. Please enter a valid integer.")
except ZeroDivisionError:
    print("Division by zero is not allowed.")
except Exception as e:
    print(e)
else:
    print("The result is:", result)
## If I want to print the result only when there is no exception, I can use else block.
## else will execute only if there is no exception raised in the try block.
## If any exception is raised, the else block will be skipped.
```

```python
## try, except, else, finally block

try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ValueError:
    print("Invalid input. Please enter a valid integer.")
except ZeroDivisionError:
    print("Division by zero is not allowed.")
except Exception as e:
    print(e)
else:
    print("The result is:", result)
## finally block will always execute, regardless of whether an exception occurred or not.
finally:
    print("Execution completed.")
```

### Real World Example

We connect to DB and while working with data, some exception occurs. Then we can handle that in some except block.
But we always need to close the connection to the DB everytime, we can write the code to close the connection in the finally block.

```python
### File Handling and Exception Handling

try:
    file = open("filename.txt", "r")
    content = file.read()
    print(content)
except FileNotFoundError:
    print("File not found. Please check the file path.")
finally:
    if 'file' in locals() and not file.closed:
        file.close()
        print("File closed.")
```

```python
### File Handling and Exception Handling
## When file does not exist
try:
    file = open("filename1.txt", "r")
    content = file.read()
    print(content)
except FileNotFoundError:
    print("File not found. Please check the file path.")
finally:
    if 'file' in locals() and not file.closed:
        file.close()
        print("File closed.")

## AS file does not exist here, the except block will execute and in finally block, since file is not opened, it will not try to close it.
## Here, the file variable is never created, locals() does not find it, so it does not try to close it.
```

## Extra Learnings

### 1. Exception vs Error

Exception: Exceptions are events that disrupt the normal flow of a program during execution.\
Handling Mechanism: They are _designed to be caught and handled programmatically_, allowing the program to potentially recover or gracefully terminate.\
Nature: Exceptions specifically refer to runtime events that can be anticipated and handled.

Error: Errors represent serious problems that typically prevent the program from running or cause it to crash unexpectedly.\
These are often _unrecoverable_ within the program's normal flow.\
Handling Mechanism: Require fixing the code or system issues.\
Nature: Errors can be syntax-related (preventing compilation/interpretation) or runtime issues.\
Eg. Syntax Errors, Indentation Errors, Memory Errors

### 2. locals() function

The locals() function in Python is a built-in function that returns a dictionary representing the current local symbol table. This symbol table contains all the local variables, functions, and other objects defined within the current scope, along with their corresponding values or references. \
It’s often used for debugging, introspection, or dynamically working with variable names and values.\
Key characteristics of locals():\
Returns a dictionary: The output of locals() is always a dictionary where keys are the names of local variables and values are their respective data.\
Reflects the local scope: When called inside a function, it provides information specific to that function's local scope, including its parameters and any variables defined within it. When called at the module level (outside any function), it returns the global symbol table, which is equivalent to globals().\
Read-only in functions (mostly): While you can inspect the local variables using locals(), _directly modifying the dictionary returned by locals() within a function generally does not affect the actual local variables._ This is because, in CPython, the local variables are stored in an optimized way, and _the dictionary returned by locals() is a copy or a view that doesn't directly map back to the variable storage._ However, in the global scope, modifications to the locals() dictionary can affect global variables.\
No parameters: The locals() function takes no arguments.

```python
## Example of locals()

def my_function(param1, param2):
    local_var = 10
    another_var = "hello"
    print(locals())

my_function(5, "world")

# Output: {'param1': 5, 'param2': 'world', 'local_var': 10, 'another_var': 'hello'}
```

### 3. Using 'with', which handles closing automatically

In the following code, no need to use finally, no need for locals()

```python
try:
    with open("filename1.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("File not found. Please check the file path.")
```

```python
try:
    with open("filename.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("File not found. Please check the file path.")
```
