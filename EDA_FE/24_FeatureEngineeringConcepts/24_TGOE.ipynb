## Target Guided Ordinal Encoding

The encoding of categorical variable into numerical variable is based on its relationship (mean or median) with the target variable.
Used when we have a categorical variable with a large number of unique categories and they are important for the model.
We rank categories from lowest to highest impact on the target.

```python
import pandas as pd

## Create DataFrame with categorical and a target variable
df = pd.DataFrame({
    'city': ['New York', 'London', 'Paris', 'Tokyo', 'New York', 'Paris'],
    'price': [200, 180, 300, 250, 180, 320]
})
df
```

```python
# Finding mean prices for each city

mean_price = df.groupby('city')['price'].mean()
mean_price.to_dict()
```

```python
df['city_encoded'] = df['city'].map(mean_price.to_dict())
```
