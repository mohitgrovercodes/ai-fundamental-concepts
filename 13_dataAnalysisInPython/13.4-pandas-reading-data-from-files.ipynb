## Reading Data From Different Sources

Using in-built functions in Pandas

```python
import pandas as pd
from io import StringIO

## Read JSON data into DataFrame
data = '{"employee_name": "Rohit", "email": "ro@hello.com", "job_title": [{"titile1": "Data Scientist"}, {"title2": "ML Engineer"}]}'
df = pd.read_json(StringIO(data))
df
```

```python
## Convert DataFrame back to JSON
json_data = df.to_json()
print("This created as key-value for each column.\n", json_data)
```

```python
## Using orient parameter to change the format with 'index'
json_data_index = df.to_json(orient='index')
print("\nThis created as key-value for each row (index).\n", json_data_index)
```

```python
## Orient options include 'split', 'records', 'index', 'columns', and 'values'.
json_data_records = df.to_json(orient='records')
print("\nThis created as list of records (dictionaries) for each row.\n", json_data_records),
```

A "pickle file" refers to a file created by the Python pickle module, which is used for serializing and deserializing Python objects.


Serialization (Pickling):\
This process converts a Python object hierarchy (like a list, dictionary, or a custom class instance) into a byte stream.\
This byte stream can then be written to a file, stored in a database, or transmitted over a network.
The act of converting an object into a byte stream is called "pickling."\
Deserialization (Unpickling):\ 
This is the inverse operation, where a byte stream (read from a file or received over a network) is converted back into a Python object hierarchy.\
The act of converting a byte stream back into an object is called "unpickling."\
Key characteristics of pickle files:\
Python-specific: Pickle is designed specifically for Python objects and is not typically compatible with other programming languages.\
Binary format: Pickle files store data in a binary format, making them efficient for storing complex Python objects.\
Versatility: Pickle can serialize almost any Python object, including functions, classes, and complex nested data structures. This is a key advantage over formats like JSON or XML, which have limitations in representing certain Python objects.\ 
Security concerns: A significant security consideration with pickle is that unpickling data from untrusted sources can lead to the execution of arbitrary code, as the pickle format can include instructions for reconstructing objects, including calling functions. Therefore, never unpickle data from untrusted sources.


Common uses of pickle files:
Saving and loading application state: Persisting the state of a program between sessions.\
Caching computations: Storing the results of expensive computations to avoid re-running them.\
Persisting machine learning models: Saving trained machine learning models for later use without needing to retrain them.\
Transmitting Python objects: Sending Python objects between different processes or systems within a Python environment.
