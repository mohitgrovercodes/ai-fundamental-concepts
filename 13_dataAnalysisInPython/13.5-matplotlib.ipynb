
## Data Visualization With Matplotlib

Matplotlib is a powerful plotting library for Python that enables the creation of static, animated, and interactive visualizations. It is widely used for data visualization in data science and analytics. In this lesson, I have learnt the basics of Matplotlib, including creating various types of plots and customizing them.

Very handy when we work in EDA and Feature Analysis

```python
import matplotlib.pyplot as plt
print("pyplot is a sub-module within the matplotlib library in Python, primarily used for creating static, animated, and interactive visualizations. It offers a MATLAB-like interface, simplifying the process of generating various types of plots and charts.")
```

### LinePlot

```python
x= [1,2,3,4,5]
y = [1,4,9,16,25]

# Create a Lineplaot
plt.plot(x,y)
```

```python
# Create a Lineplaot
plt.plot(x,y)

## basic c=components
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Basic line plot')
plt.show
```

```python
## Customise the plot

plt.plot(x,y,color='red',linestyle='-.',marker='o',linewidth='2',markersize=9)
plt.grid(True)
```

```python
# Multiple LinePlots
## Sample data
x = [1,2,3,4,5]
y1 = [1,4,9,16,25]
y2 = [1,2,3,4,5]

# Parent figure to create subplots
plt.figure(figsize=(9,5))

# Plot the sub figure
plt.subplot(2,2,1)   #rowsno (2), columnsno (2), subplotposition
plt.plot(x,y1,color='red')
plt.title('Plot 1')

# Second plot
plt.subplot(2,2,2)
plt.plot(x,y2,color='green')
plt.title('Plot 2')

plt.subplot(2,2,3)
plt.plot(y1,x,color='yellow')
plt.title('Plot 3')

plt.subplot(2,2,4)
plt.plot(y1,y2,color='purple')
plt.title('Plot 4')

```

### Bar Plots (Building-like)

```python
categories = ['A','B','C','D','E']
values = [3,5,4,8,4]

#Create bar plot
plt.bar(categories,values,color='skyblue')
```

##  Histogram

Histograms are used to represent the distribution of a dataset. They divide the data into bins and count the number of data points in each bin.

```python
data = [1,2,4,3,6,2,9,8,1,7,4,5]

# create histogram
plt.hist(data,bins=5,color='orange',edgecolor='black')
```

## Scatterplot

```python
x = [1,2,3,4,5]
y = [5,4,3,1,3]

plt.scatter(x,y,color='red')
```

## Piechart

```python
labels = ['A','B','C','D','E']
size = [25,30,15,10,20]
color = ['gold','yellowgreen','teal','lightcoral','lightskyblue']
explode = (0.2,0,0,0,0) #move out the piece
# Create Pie Chart
plt.pie(size,labels=labels,colors=color,autopct='%1.1f%%',shadow=True,explode=explode)
```

## Visualize Sales Data

```python
import pandas as pd
sales_data_df = pd.read_csv('sales1_data.csv')
sales_data_df.head()
```

```python
# Print total sales by product

total_sales = sales_data_df.groupby('Product')["Sales"].sum()
print('Data type is:',type(total_sales))
print(total_sales)
print('\nResult is a Series')
```

```python
total_sales_bad = sales_data_df.groupby('Product').sum('Sales')
print(type(total_sales))
print(total_sales)
print('\nThis is incorrect! Sum ignpres the argument (treats it like axis). It sums up all the numeric columns.\nThe argument "Sales" is ignored.\n Result is a DF')
```

```python
total_sales.plot(kind='bar',color='teal')
plt.xlabel("Product")
plt.ylabel("Sales")
```

```python
# Plot sales plot over time

print(sales_data_df.groupby('Date')['Sales'].sum())
```

```python
# Use reset_index() before plotting and after using groupby() as griupby sets selected column as index
sales_data_df.groupby('Date')['Sales'].sum().reset_index()
plt.plot(sales_data_df['Date'],sales_data_df['Sales'])
```
