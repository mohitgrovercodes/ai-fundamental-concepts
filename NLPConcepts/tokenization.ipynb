## Tokenization

Process of splitting text into smaller units:
1. Sentence Tokenization: Breaking corpus (paragraph) into documents (sentences).
2. Word Tokenization: Document into words.

Use of NLTK library...(open source)

```python
corpus = '''Hi! Welcome to my learnings. Let's learn NLP together.
Today is an amazing day.'''
```

```python
corpus
```

```python
print(corpus)
```

```python
## Sentence Tokenization

import nltk
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize
documents = sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
## Word Tokenization
## Document -> Words
## Paragraph -> Words

from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)  ##Punctuation also treated as word
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize

wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer

tokenizer_new = TreebankWordTokenizer()
```

```python
tokenizer_new.tokenize(corpus)
```
