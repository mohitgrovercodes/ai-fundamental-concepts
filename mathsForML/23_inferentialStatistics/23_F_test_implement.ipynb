## Implementing Variance Ratio Test (F Test) with Python

The number of bulb units produced by two workers is given. We need to find our which worker is more efficient or they are similar.

```python
# evidence to say that the two population variances are not equal

worker1 = [18,19,22,25,27,28,41,45,51,55]
worker2 = [14,15,15,17,18,22,25,25,27,34]
```

```python
## Calculating F Test

import numpy as np

var1 = np.var(worker1)
var2 = np.var(worker2)

f_test = var1/var2 # To compare with Critical Value later
```

```python
f_test
```

```python
# Defgree of Freedom

df1 = len(worker1) - 1
df2 = len(worker2) - 1
```

```python
# Significance Value (given by domain expert)
sig_val = 0.05
```

```python
# Performing F-test

import scipy.stats as stat

critical_val = stat.f.ppf(q=1-sig_val,dfn=df1,dfd=df2)
```

```python
if f_test > critical_val:
    print("Reject the Null Hypothesis")
else:
    print("We fail to reject the null hypothesis.")
```
