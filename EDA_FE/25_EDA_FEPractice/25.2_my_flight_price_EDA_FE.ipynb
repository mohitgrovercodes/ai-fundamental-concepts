### Extra Learnings

1

Always Convert to DateType and then deal with it using date accessor .dt
Robust & safe

Handles invalid dates → NaT

Format-agnostic

Works even if format slightly varies

Built-in validation

Pandas ensures real calendar dates

Extensible

Weekday, week number, quarter, weekend flags

ML-friendly

Cleaner, consistent numeric features

2

duplicated() gives exact duplicates always. We can find logical (complex) duplicates based on multiple features using

df.drop_duplicates(
    subset=["Airline", "Date_of_Journey", "Source", "Destination", "Dep_Time", "Price"],
    inplace=True
)

3

Instead of hard rule based parsing, use regex. Or even better logics::

import numpy as np
import pandas as pd

def convert_duration(duration):
    if pd.isna(duration):
        return np.nan

    duration = duration.strip()
    hours = 0
    minutes = 0

    if 'h' in duration:
        hours = int(duration.split('h')[0])

    if 'm' in duration:
        minutes = int(duration.split('m')[0].split()[-1])

    return hours * 60 + minutes

df['Duration_mins'] = df['Duration'].apply(convert_duration)
