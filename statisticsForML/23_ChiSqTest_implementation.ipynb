## Implementation of Chi Square Test Goodness of Fit

The warden of a hostel was asked how many average hours do students study per day of the week (expected).
Number of hours they were actually studying (observed).
Can we say warden gave the right info?
Use significance value = 0.05


```python
import numpy as np
import scipy.stats as stat
```

```python
## Number of hours students study in a week on daily basis

expected = [8,6,7,9,6,9,7]
observed = [7,8,6,9,9,6,7]
```

```python
## Chi-Square Goodness of Fit test

chi_sq_test_stats, p_value = stat.chisquare(observed,expected)
```

```python
chi_sq_test_stats, p_value
```

```python
## Find Critical Value

sig_val = 0.05
df = len(expected) - 1
critical = stat.chi2.ppf(1-sig_val,df)
critical
```

```python
if chi_sq_test_stats > critical:
    print("We reject the null hypothesis.")
else:
    print("we fail to reject the null hypothesis.")
```

Summation of observed and expected data must be equal.
