# Handling Outliers

## 5 Number Summary

1 Minimum Value -0% <br>
2 Q1 - 25%<br>
3 Median -50%<br>
4 Q3 -75%<br>
5 Maximum -100%

```python
import numpy as np
marks = [45,53,60,399,49,80,44,52,58,48,210,48]
```

```python
np.percentile(marks,25)
```

```python
## [Lower Fence <----> Higher Fence]
Q1 = np.percentile(marks,[25])
print(Q1)
```

```python
minimum, Q1,Q2,Q3,maximum=np.quantile(marks,[0,0.25,0.50,0.75,1])
```

```python
maximum
```

```python
# IQR = Interquartile Range
IQR=Q3-Q1
print(IQR)
```

```python
lower_fence=Q1-1.5*(IQR)
higher_fence=Q3+1.5*(IQR)
```

```python
lower_fence,higher_fence
```

```python
outliers =[]
for i in marks:
  if i<lower_fence or i>higher_fence:
    outliers.append(i)
    print("Outlier value")
  else:
    print("Not an outlier")
```

```python
outliers
```
