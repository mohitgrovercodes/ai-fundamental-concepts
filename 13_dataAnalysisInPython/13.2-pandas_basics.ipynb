
## Pandas-DataFrame And Series

Pandas stands for Python Data Analytics.Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning.\
It provides two primary data structures: Series and DataFrame.\
A Series is a one-dimensional array-like object, while a DataFrame is a _two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes_ (rows and columns).

We can convert 1D numpy array into a Pandas Series
And 2D numpy matrics into a Pandas DataFrame

```python
## Import Pandas libraries
 
import pandas as pd
```

```python
## Series - A Pandas Series is a one-dimensional array-like object that can hold any data type. It is similar to a column in a table. It is labled.

import pandas as pd

data=[1,2,3,4,5]
series=pd.Series(data)
print("Series \n",series)
print(type(series))
## The Series is indexed default (0,1,2,3,4)
```

```python
## Create a series from Dict

dic = {
    'a': 1,
    'b': 2,
    'c': 3
}
series_dic = pd.Series(dic)
print(series_dic)
## The keys become index and values become Series values
```

```python
data = [12,34,556]
keys = ['a','b','c']
pd.Series(data, index=keys)
```

```python
## we can rename the index
dic = {
    'a': 1,
    'b': 2,
    'c': 3
}
pd.Series(data, index=['x','y','z'])
```

```python
## DataFrame
## Create DataFrame from dictionaries

data = {
    'Name': ['Mohit', 'Love', 'Surat'],
    'Age': [23,45,24],
    'City': ['Delhi','LA','SF']
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## Can convert df to nparray
import numpy as np
np.array(df)
```

```python
## Accessing the data from df

data = {
    'Name': ['Mohit', 'Love', 'Surat'],
    'Age': [23,45,24],
    'City': ['Delhi','LA','SF']
}

df = pd.DataFrame(data)
df
```

```python
# We can call any column, it actually behaves like a Series
df['Name']
```

```python
type(df['Name'])
```

```python
## Accessing multiple columns
df[['Name','Age']]
```

```python
## Label-based Indexing
## Access row using .loc  when row and column names are known
df.loc[0,'Age']
```

```python
## Another method to access column using iloc (indexloc) opr similar to list opr
print(df.iloc[0])
```

```python
df.iloc[0][0]
```

```python
## Accessing a specified element (name of labels)
df.at[1,'Name'] # pyright: ignore[reportUndefinedVariable]
```

```python
## Accessing a specifoed element using iat (index of element)
df.iat[2,2] # pyright: ignore[reportUndefinedVariable]
```
