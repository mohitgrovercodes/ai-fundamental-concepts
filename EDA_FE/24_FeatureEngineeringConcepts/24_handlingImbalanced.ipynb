## Handling Imbalanced Datasets

An imbalanced dataset is one where the classes in the target variable are not represented equally.

Can be dealt with two basic methods:
1. Upsampling
2. Downsampling

```python
import numpy as np
import pandas as pd

## Set random seed for reproducibility
np.random.seed(142)

# Create a DataFrame with two categories

n_samples = 10000
class_0_ratio = 0.85
n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0,n_class_1
```

```python
## Create DataFrame with impbalanced Dataset

class_0 = pd.DataFrame({
    'f1': np.random.normal(loc=0, scale=1,size=n_class_0),
    'f2': np.random.normal(loc=0,scale=1,size=n_class_0),
    't': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'f1': np.random.normal(loc=2, scale=1,size=n_class_1),
    'f2': np.random.normal(loc=2,scale=1,size=n_class_1),
    't': [1] * n_class_1
})
```

```python
df = pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.head()
```

```python
df.sample(10)
```

```python
df['t'].value_counts()
```

### Upsampling Technique

```python
## Create 2 DataFrames

df_minority=df[df['t']==1]
df_majority=df[df['t']==0]
```

```python
# Upsampling with replacement of values
from sklearn.utils import resample
df_minortiy_upsampled = resample(df_minority, replace=True,  ## Sample with replacement
        n_samples=len(df_majority),
        random_state=22
)
```

```python
df_minortiy_upsampled.shape
## Shows we have created more datapoint with value = 1
```

```python
df_upsampled = pd.concat([df_majority,df_minortiy_upsampled])
```

```python
df_upsampled.sample(10)
```

```python
## Balanced DataSet
df_upsampled['t'].value_counts()
```

## Downsampling

```python
## Creating the same DataFrame again

## Create DataFrame with impbalanced Dataset

class_0 = pd.DataFrame({
    'f1': np.random.normal(loc=0, scale=1,size=n_class_0),
    'f2': np.random.normal(loc=0,scale=1,size=n_class_0),
    't': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'f1': np.random.normal(loc=2, scale=1,size=n_class_1),
    'f2': np.random.normal(loc=2,scale=1,size=n_class_1),
    't': [1] * n_class_1
})

df = pd.concat([class_0,class_1]).reset_index(drop=True)

# Check the class distribution
print(df['t'].value_counts())
```

```python
## Create 2 datasets

## Create 2 DataFrames

df_minority=df[df['t']==1]
df_majority=df[df['t']==0]
```

```python
# Downsampling with replacement of values
from sklearn.utils import resample
df_majority_downsampled = resample(df_majority, replace=False,  ## Don't want to replace as we want to reduce the datapoints
        n_samples=len(df_minority),
        random_state=22
)
```

```python
df_majority_downsampled.shape
```

```python
df_downsamped = pd.concat([df_minority,df_majority_downsampled])
df_downsamped['t'].value_counts()
```
