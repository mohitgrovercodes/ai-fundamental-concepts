## Working With File Paths

When working with files in Python, handling file paths correctly is crucial to ensure your code works across different operating systems and environments. \
Python provides several modules and functions for working with file paths effectively.

```python
## Using the os module

import os
cwd = os.getcwd()
print(f"The Current working Directory is {cwd}")
```

```python
## Creating a new directory

new_dir = 'package_dir'
os.mkdir(new_dir)
print(f"Created new directory: {new_dir}")
```

```python
## Listing files and directories in current directory
## Important conceppt
items = os.listdir('.')  ## '.' refers to current directory
print("Items in current directory:")
for item in items:
    print(item)
```

```python
## Joining paths - Very important concept
## This is OS independent way of joining paths

dir_name = 'folder'
file_name = 'data.txt'
full_path =  os.path.join(dir_name, file_name)
print(f"Relative path: {full_path}")
```

```python
dir_name = 'folder'
file_name = 'data.txt'
full_path =  os.path.join(os.getcwd(),dir_name, file_name)
print(f"Full path: {full_path}")
```

```python
## Checking if a file exists on a path\

path = "example1.txt"
if os.path.exists(path):
    print(f"The file {path} exists.")
else:
    print(f"The file {path} does not exist.")
```

```python
## Checking if a path is a file or directory

path = "example1.txt"
if os.path.isfile(path):
    print(f"{path} is a file.")
elif os.path.isdir(path):
    print(f"{path} is a directory.")
else:
    print(f"{path} does not exist.")            
```

```python
## Getting the absolute path of a file

relative_path = 'example1.txt'
absolute_path = os.path.abspath(relative_path)
print(f"Absolute path: {absolute_path}")    
```

```python
## Getting file size
file_path = 'example1.txt'
if os.path.exists(file_path):
    size = os.path.getsize(file_path)
    print(f"Size of {file_path} is {size} bytes.")
```

```python
## Removing a file
file_to_remove = 'temp_file.txt'
if os.path.exists(file_to_remove):
    os.remove(file_to_remove)
    print(f"Removed file: {file_to_remove}")
```

```python
## Removing a directory
dir_to_remove = 'package_dir'   
if os.path.exists(dir_to_remove):
    os.rmdir(dir_to_remove)
    print(f"Removed directory: {dir_to_remove}")
```
