## Data Visualization With Seaborn
Seaborn is a Python visualization library based on Matplotlib that provides a high-level interface for drawing attractive and informative statistical graphics. Seaborn helps in creating complex visualizations with just a few lines of code. In this lesson, we will cover the basics of Seaborn, including creating various types of plots and customizing them.

```python
import seaborn as sns
```

```python
## Basic plotting with seaborn
tips = sns.load_dataset('tips')
tips
```

```python
sns.scatterplot(x='total_bill',y='tip',data=tips)

import matplotlib.pyplot as plt

plt.title("Scatterplot of bill and tips")
plt.show()
```

```python
## Line Plot

sns.lineplot(x='size',y='total_bill',data=tips)
plt.title("Line plot of tips ds")
plt.show()
```

### Categorical Plots using categorical variables

```python
## Bar Plot

sns.barplot(x='day',y='total_bill',data=tips)
plt.show()
```

```python
## Box Plot
sns.boxplot(x='day',y='total_bill',data=tips)
plt.show()
```

```python
## Violin Plot
## Distribution of datapoints

sns.violinplot(x='day',y='total_bill',data=tips)
plt.show()
```

```python
## Histograms
sns.histplot(tips['total_bill'],bins=8, kde=True)
## kde --> Kernel Density Estimator smoothens the hist, shows distribution
```

```python
## KDE Plot

sns.kdeplot(tips['total_bill'],fill=True)
```

```python
## Pair Plot
## to see relation between all the NUMERICAL variables
sns.pairplot(tips)
```

```python
## Heatmap

mycorr = tips.corr('tip','total_bill','size')
```

```python
mycorr = tips[['total_bill','tip','size']].corr()
mycorr
```

```python
## Heqtmap gives a colorful visualization
sns.heatmap(mycorr,annot=True,cmap='coolwarm')
```

```python
import pandas as pd

df = pd.read_csv('sales_data.csv')
df.head()
```

```python
## Plot total sales by price
plt.figure(figsize=(10,6))
sns.barplot(x='Product Category',y='Total Revenue',data=df,estimator=sum)  ##estimator feature to do agg not agg()
plt.title("Total Sales vs Producct")
plt.xlabel('Category')
plt.ylabel('Revenue')
plt.show()
```

Seaborn provides:

cleaner color palettes

more readable gridlines

better spacing

better handling of categorical data

visually appealing defaults

3\
Seaborn gives you complex statistical plots with 1 line\
eg. Boxpolt, violinplot, regression plot

4\

Example: automatically computes mean + confidence intervals

5\
Seaborn simplifies multi-variable and categorical plotting

Seaborn excels where Matplotlib becomes painful:

grouped bar charts

facet grids (small multiples)

heatmaps

categorical scatterplots

distribution plots
