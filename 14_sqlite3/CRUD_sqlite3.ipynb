## SQL and Sqlite3



SQL (Structured Query Language) is a standard language for managing and manipulating relational databases.\
SQLite is a self-contained, serverless, and zero-configuration database engine that is widely used for embedded database systems. I have covered the basics of SQL and SQLite, including creating databases, tables, and performing various SQL operations(CRUD).\
In Python3, SQLite3 comes pre installed.

```python
import sqlite3
```

```python
## Connect to SQLite DB
mydb = sqlite3.connect('mydb.db')
mydb  # created without any config
```

```python
## Create cursor to iterate though the db, tables, rows
cursor = mydb.cursor()
```

```python
## Create table
cursor.execute('''
CREATE TABLE IF NOT EXISTS Employees(
    id INT PRIMARY KEY,
    name TEXT NOT NULL,
    age INT,
    department TEXT
    )
''')

## To execute finally, we need to commit the changes
mydb.commit()
```

```python
cursor.execute('''
SELECT * FROM Employees
''')
```

```python
## Insert data in SQLite3 table

cursor.execute('''
    INSERT INTO Employees(name, age, department)
    VALUES
        ("Mohit",30,"Data Scientist"),
        ("Harish",34,"HR"),
        ("Ramu",28,"Backend Eng")
'''
)

mydb.commit()
```

```python
## Querying the data from the table

cursor.execute('SELECT * FROM Employees') #tells sqlite to prepare the results and this stores them in the cursor's result buffer
rows = cursor.fetchall()

'''
Others are:
fetchone(), fetchmany(n)
'''
## To print the results
for row in rows:
    print(row)

# Here we did not commit() as no changes made
```

```python
## Update data in table
with mydb: # with keyword, no need to commit
    cursor.execute('''
    UPDATE Employees
    SET age=34
    WHERE name="Ramu"
    ''')
```

```python
## Delete the data from table
with mydb:
    cursor.execute('''
    DELETE FROM Employees
    WHERE name="Ramu"
    ''')
```

### Practice - Working with Sales DB


```python
## Connect with SQLite3 DB
conn = sqlite3.connect('sales_data.db')
```

```python
## Create cursor
cursor = conn.cursor()

## Create table
cursor.execute('''
CREATE TABLE IF NOT EXISTS Sales(
    id INT PRIMARY KEY,
    date TEXT NOT NULL,
    product TEXT NOT NULL,
    sales INT NOT NULL,
    region TEXT
    )
''')

## insert multiple entries
sales_entries = [
    ('2024-02-19','Trimmer',190,'West'),
    ('2024-04-19','Oven',148,'West'),
    ('2024-02-19','JellyCake',300,'East'),
    ('2024-02-19','Food',190,'Central'),
    ('2024-02-19','Refrigirator',15,'South'),
]

cursor.executemany('''
INSERT INTO Sales(date,product,sales,region)
VALUES(?,?,?,?)
''',sales_entries)

##  Adding ? in values prevent SQLInjection

conn.commit()
```

```python
## At last we close the connection
conn.close()
```

```python
## Can't do any more actions
cursor.execute('SELECT * FROM Sales')
```

An embedded database is a lightweight database that is tightly integrated into an application, running locally within the same process and not requiring a separate server.\
This architecture makes it ideal for mobile, IoT, and other resource-constrained or offline environments where local access and low latency are critical (no network latency).\

Common use cases:\
1 Mobile apps: Store user preferences, and offline content.\
2 IoT: Theses are often disconnected or have limited connectivity.\
3 Edge computing: Performing local data processing on devices at the "edge" of the network.\
4 Caching: Acting as a high-speed cache for an application to improve read and write speeds. \
5 Desktop and specialized software: Enabling applications to store data locally without relying on a central server.

2

Edge can include retail stores, factory floors, vehicles, remote offices, or even individual devices.
