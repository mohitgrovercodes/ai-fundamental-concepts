```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification # Creates binary or multiclass classification

X, y = make_classification(
    n_samples=10000,
    n_features= 2,
    n_clusters_per_class=1,
    weights= [0.94], #Imbalance
    random_state=10,
    n_redundant=0 # Overwrite
)

## gives 2 arrays as output
```

```python
## Create two DF and joining them
df1 = pd.DataFrame(X,columns=['f1','f2'])
df2 = pd.DataFrame(y,columns=['target'])
df = pd.concat([df1,df2],axis=1)
df.head()
```
