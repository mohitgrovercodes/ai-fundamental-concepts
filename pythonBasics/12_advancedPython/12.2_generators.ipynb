
## Generators
Generators are a simpler way to create iterators. They use the yield keyword to produce a series of values lazily, which means they generate values on the fly and do not store them in memory thus managing memory efficiently.\
Generators can be said as shortcut iterators.\
They remember their state.\

```python
def cube(num):
    for i in range(3):
        return pow(i,3)
```

```python
cube(3)
## Returns the first value only
```

```python
## With yield keyword, the value gets saved and return it  
def cube(num):
    for i in range(num):
        yield pow(i,3)
```

```python
type(cube(3))
## gives Genrator object
```

```python
for i in cube(3):
    print(i)
```

```python
a = cube(3)
a
```

```python
next(a)
```

```python
##We can return multiple values by using yield in generator
def my_gen():
    yield 1
    yield 2
    yield 3
```

```python
gen_var = my_gen()
gen_var
```

```python
next(gen_var)
```

```python
for val in gen_var:
    print(val)
```


### Practical Example: Reading Large Files
Generators are particularly useful for reading large files because they allow you to process one line at a time without loading the entire file into memory.

```python
## Practical Example: Reading Large file
def read_file(file_path):
    with open(file_path,'r') as f:
        for line in f:
            yield line
```

```python
file_path = 'para.txt'
for line in read_file(file_path):
    print(line.strip())
```
