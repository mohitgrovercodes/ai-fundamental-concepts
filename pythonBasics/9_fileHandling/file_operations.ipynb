## File Operation- Read And Write Files

File handling is a crucial part of any programming language. Python provides built-in functions and methods to read from and write to files, both text and binary. This lesson will cover the basics of file handling, including reading and writing text files and binary files.\

```python
## 1 READING FROM FILES
## (1A) Reading Read a Whole File
## This script reads the entire content of a file named 'example.txt' and prints it to the console.
## creates file object in read mode

with open('example.txt','r') as file:
    content=file.read()
    print(content)

## Gives error if file not found
```

```python
## Now made a txt file named example.txt in the same directory
with open('example.txt','r') as file:
    content=file.read()
    print(content)
```

```python
## (1B) Read a file line by line

with open('example.txt','r') as file:
    for line in file:
        print(line)
## This code adds next line after each line because of existing newline character
```

```python
# strip() to remove extra newlines and spaces

with open('example.txt','r') as file:
    for line in file:
        print(line.strip())
```

```python
## Reading specific number of characters
with open('example.txt','r') as file:
    content=file.read(10)  ## reads first 10 characters
    print(content)
```

```python
## (2) WRITING TO FILES
## This script writes multiple lines of text to a file named 'output_file.txt'.
## (2A) Writing to a file by overwriting existing content

with open('output_file.txt','w') as file:
    file.write("Hello, World!\n")
    file.write("This is a test file.\n")
    file.write("Writing to files in Python is easy.\n")

with open('output_file.txt','w') as file:
    file.write("Overwritten line 1.\n")
    file.write("Overwritten line 2.")
```

```python
## (2B) Writing without overwriting/Appending to a file

with open('output_file.txt','a') as file:
    file.write("\nI am not overwriting existing content.")
    file.write("\nJust appended line.")
```

```python
## (2C) Writing a list of lines to a file (multiple lines at once)

lines = ["First line.\n", "Second line.\n", "Third line.\n"]
with open('output_file.txt','a') as f:
    f.writelines(lines)

```

```python
## 3 WORKING WITH BINARY FILES
## This script demonstrates how to write binary files in Python. We write in bytes.
## (3A) Writing binary data to a file

with open('binary_file.bin','wb') as bin_file:
    bin_data = b'\x00\xFF\x7A\x3C\x5D'
    bin_file.write(bin_data)
```

```python
## (3B) Reading binary data from a file

with open('binary_file.bin','rb') as bf:
    bf_data = bf.read()
    print(bf_data)
```

```python
## Read content from a source text file and write it to a destination file

with open('example.txt','r') as file:
    content = file.read()

with open('destination_file.txt','w') as dest_file:
    dest_file.write(content)
```

```python
## Read a text file and count the number of lines, words, and characters

with open('example.txt','r') as file:
    lines = [line for line in file]
    line_count = len(lines)
    word_count = sum(len(line.split()) for line in lines)
    char_count = sum(len(line) for line in lines)
    print(f"Lines: {line_count}, Words: {word_count}, Characters: {char_count}")
```

```python
## Above logic with functions

def count_file_stats(file_path):
    with open(file_path,'r') as file:
        lines = [line for line in file]
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

lines, words, chars = count_file_stats('example.txt')
print(f"Lines: {lines}, Words: {words}, Characters: {chars}")
```

The w+ mode in Python is used to open a file for both reading and writing. If the file does not exist, it will be created. If the file exists, its content is truncated (i.e., the file is overwritten).

```python
## Writing and then reading files (w+ mode)

with open('example1.txt','w+') as file:
    file.write("This is a new file created with w+ mode.\n")
    file.write("We can write and then read from it.\n")
    # Move the cursor back to the beginning of the file
    file.seek(0)
    ## Read the content from beginning
    content = file.read()
    print(content)
```

### All the modes in file handling in Python

BASIC MODES

1 'r' (Read Mode)
Opens a file for reading.\
The file pointer is placed at the beginning of the file. \
Raises a FileNotFoundError if the file does not exist.\
This is the default mode if no mode is specified.\

2 'w' (Write Mode)
Opens a file for writing.\
If the file exists, its contents are truncated (erased).\
If the file does not exist, a new file is created. \
The file pointer is placed at the beginning of the file. \

3 'a' (Append Mode)
Opens a file for appending.\
If the file exists, new data is written to the end of the file without overwriting existing content.\
If the file does not exist, a new file is created. \
The file pointer is placed at the end of the file. \

4 'x' (Exclusive Creation Mode)
Opens a file for exclusive creation.\
If the file already exists, a FileExistsError is raised.\
If the file does not exist, a new file is created for writing.\

COMBINED MODES

5 'r+' (Read and Write Mode):
Opens a file for both reading and writing.\
The file must exist; otherwise, a FileNotFoundError is raised.\
The file pointer is placed at the beginning of the file. \

6 'w+' (Write and Read Mode):
Opens a file for both writing and reading.\
If the file exists, its contents are truncated.\
If the file does not exist, a new file is created. \
The file pointer is placed at the beginning of the file. \

7 'a+' (Append and Read Mode):
Opens a file for both appending and reading.\
If the file exists, new data is appended to the end.\
If the file does not exist, a new file is created. \
The file pointer is initially at the end for writing, but can be moved for reading.\

BINARY MODE:\
You can combine any of the above modes with 'b' to open the file in binary mode. This is used for handling non-text files like images, audio, or executable files.\
'rb': Read in binary mode.\
'wb': Write in binary mode.\
'ab': Append in binary mode.\
'r+b': Read and write in binary mode.\

TEXT MODE (DEFAULT):\
You can also explicitly specify 't' for text mode, although it's the default for most modes.\
'rt': Read in text mode (same as 'r').\
'wt': Write in text mode (same as 'w').
