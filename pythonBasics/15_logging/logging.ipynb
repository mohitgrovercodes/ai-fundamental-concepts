# Python Logging

Logging is a crucial aspect of any application, providing a way to track events, errors, and operational information. Python's built-in logging module offers a flexible framework for emitting log messages from Python programs. In this lesson, we will cover the basics of logging, including how to configure logging, log levels, and best practices for using logging in Python applications.

Logging is super important for any end-to-end project develpment and maintenance wrt execution. Useful for debugging the app.

```python
## Configure the basic logging settings

import logging

## Whenever we import logging, we need to configure the basic logging settings

## wrt level of logging, there are different levels like DEBUG, INFO, WARNING, ERROR, CRITICAL
## as we are just developing code, we can set it to DEBUG to see all the logs
logging.basicConfig(level=logging.DEBUG)

## log messagaes can be added using logging.debug(), logging.info(), logging.warning(), logging.error(), logging.critical(). They have different severity levels.
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

### Log Levels
Python's logging module has several log levels indicating the severity of events. The default levels are:

* DEBUG: Detailed information, typically of interest only when diagnosing problems.
* INFO: Confirmation that things are working as expected.
* WARNING: An indication that something unexpected happened or indicative of some problem in the near future (e.g., ‘disk space low’). The software is still working as expected.
* ERROR: Due to a more serious problem, the software has not been able to perform some function.
* CRITICAL: A very serious error, indicating that the program itself may be unable to continue running.

```python
## Configuring logging 

logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)

logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

```python
## Configuring logging again
## But this does not work because basicConfig can be called only once. Subsequent calls are ignored.
## We can fix this by restarting the kernel or using force=True in basicConfig (Python 3.8+)
## Restarting the kernel is the simpler way but should avoid in production code and long running applications.
import logging

logging.basicConfig(
    filename='app.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)

logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

```python
## Again adding more logs

logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

## Problem occurred in format of logs

The real reason you're not seeing the timestamp is because:
VS Code already initialized logging BEFORE your code ran.
And Python’s logging.basicConfig() only works the first time logging is configured.
After that, any calls to basicConfig() are ignored.

In Jupyter notebooks, VS Code, and some IDEs:
The environment initializes logging internally
That means basicConfig() in your code does nothing, because logging is already configured
So the default logging format is shown (which does NOT include time/date)

Solution:
1. Force reconfigure logging using _force=True_
2. Remove existing handlers manually
import logging

for handler in logging.root.handlers[:]:
    logging.root.removeHandler(handler)

logging.basicConfig(
    level=logging.DEBUG,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
3. Restart the kernel
When you restart the kernel (in VS Code, Jupyter, or any interactive environment):

All variables are cleared
All modules are unloaded
All logging handlers are removed
Python starts fresh

The Python logging system is global and persistent for the entire kernel session.
Once handlers are added (often by VS Code / IPython automatically), calling basicConfig() again won’t override them.

Restarting the kernel resets this global state.
