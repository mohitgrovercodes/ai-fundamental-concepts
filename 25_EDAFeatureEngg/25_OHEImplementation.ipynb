## Nominal/ One Hot Encoding

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
## Create a DataFrame
df = pd.DataFrame({
    'color': ['red','green','yellow','black','white','blue']
})
```

```python
df.head()
```

```python
## Create an instance of OneHotEncoder
encoder = OneHotEncoder()
```

```python
## Perform fit and transform

encoded = encoder.fit_transform(df[['color']]).toarray()
# Creates alphabetically
```

```python
encoded_df = pd.DataFrame(encoded,columns=encoder.get_feature_names_out())  ## use this fn to output feature names
```

```python
encoded_df
```

```python
## For converting any new values

encoder.transform([['blue']]) ## creates sparse matrix
```

```python
##Convert to array

encoder.transform([['blue']]).toarray()
```

```python
## Concat

df_final = pd.concat([df,encoded_df],axis=1)
df_final
```
