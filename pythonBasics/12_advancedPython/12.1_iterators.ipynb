## Iterators Concept
Iterators are advanced Python concepts that allow for efficient looping and memory management. An iterator is an _object_ that you can loop over one item at a time. Iterators provide a way to access elements of a collection _sequentially without exposing the underlying structure._

```python
my_list = [1,2,3,5,6,8]
for i in my_list:
    print(i)
```

```python
type(my_list)
```

```python
## Creating an iterator from list
iterator = iter(my_list)
type(iterator)
```

```python
# Does not expose the underlying list unless we iterate over it
iterator
```

```python
## To iterate through all the elements, we use function called next()
## LAZY LOADING is used, only one element is loaded at a time
## So efficient memory use
next(iterator)
```

```python
iterator = iter(my_list)
```

```python
try:
    print(next(iterator))

except StopIteration:
    print("No more elements in iterator.")
```

```python
## We can use it over a string

my_str = "Strings"
str_iterator = iter(my_str)
```

```python
while True:
    try:
        print(next(str_iterator))
    except StopIteration:
        print('String over')
        break
```

### Extra Learnings

Iterable - something you can loop over (list, string, set…).\
Iterator = object that returns items one by one using next().\
for loops internally use iterators.\
Can create your own iterator by defining '__iter__()' and '__next__()' magic fn.
