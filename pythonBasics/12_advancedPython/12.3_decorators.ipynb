## Decorators
Decorators are a powerful and flexible feature in Python that allows you to modify the behavior of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code. The basics of decorators, including how to create and use them.

- function copy
- closures
- decorators

### Function Copy

```python
def welcome():
    return "This is an Advanced Paython topic"

welcome()
```

```python
wel = welcome # Create a function copy
wel
```

```python
wel()
```

```python
print(wel())

del welcome

print(wel())
## Copied func works same even after original fun is deleted
```

```python
import sys

print(sys.getrefcount(wel))
```

## Closures
Closures are functions inside a function

```python
def main_fn():
    msg = "Welcome to main"
    def sub_fn():
        print("This is sub function")
        print(msg)
        print("Learn happy!")
    return sub_fn()
## Return type of main function is sub_fn.
```

```python
main_fn()
## Gets sub_fn messages
## We can access the main_fn local variable in sub_fn also (outside itself)
```

```python
def main_fn2(msg):
    
    def sub_fn():
        print("This is sub function")
        print(msg)
        print("Learn happy!")
    return sub_fn()
```

```python
main_fn2("Hey")
```

```python
## Following fn returns the inner fn itself and doesnot call it.
def main_fn():
    msg = "Welcome to main"
    def sub_fn():
        print("This is sub function")
        print(msg)
        print("Learn happy!")
    return sub_fn
```

```python
f = main_fn()
f ##Function object -> outer fn -> inner function
```

```python
print(f.__closure__[0])
print(f.__closure__[0].cell_contents)
```

```python
f()
```

```python
## We can use other functions as functional parameters as well
def main_fn3(func):
    
    def sub_fn():
        print("This is sub function")
        func("I am argument to print.")
        print("Learn happy!")
    return sub_fn()
```

```python
main_fn3(print)
```

```python
## Any function can be passed
def main_fn3(func):
    
    def sub_fn():
        print("This is sub function")
        print(func([1,3,5,7,7]))
        print("Learn happy!")
    return sub_fn()
## Even though the main_fn3 has has finished executing, sub_fn still has access to the parameter beacause Python stores that in the referencing environment.
```

```python
main_fn3(len)
```

### Decorators

```python
## Performing closures manually
def main_fn(func):
    
    def sub_fn():
        print("This is sub function")
        func()
        print("Learn happy!")
    return sub_fn()
```

```python
def adv_python():
    print("Happy reading")
adv_python()
```

```python
main_fn(adv_python)
```

```python
## Using Decorator, we can do the above action without calling the original method.
def main_function(func):
    def sub_function():
        print("This is happening befor the function call.")
        func()
        print("This is happening after the function call.")
    return sub_function()
```

```python
@main_function
def mid_function():
    print("This is an advanced concept.")
```

```python
## Another example

def my_decorator(func):
    def wrapper():
        print("Something")
        func()
        print("Something more")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello!")
```

```python
say_hello()
```

```python
## Decorators with arguments

def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(4)
def say_hello():
    print("Hello!")
```

```python
say_hello()
```

### Extra Learnings
1

A closure occurs when an inner function “remembers” and has access to variables from its enclosing (outer) function—even after the outer function has finished executing.

2

Enclosing scope: The outer function’s scope.\
Free variable: A variable used by a function but not defined inside it.\
Closure: The combination of a function + its referencing environment.\
Referencing environment: The collection of those free variables and their values.

3

Why decorators matter (especially in AI/ML)

Decorators are powerful for:
- Logging function calls
- Measuring execution time
- Caching expensive computations
- Validating inputs
- Pre/post-processing in ML pipelines
- Automatically tracing model functions\
They're everywhere in frameworks like Flask, FastAPI, PyTorch, TensorFlow, etc.
