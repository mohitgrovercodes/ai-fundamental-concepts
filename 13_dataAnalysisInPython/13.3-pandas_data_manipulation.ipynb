## Data Manipulation using Pandas DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [24, 30, 22, 35],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
})
```

```python
df
```

```python
## Adding a new column with default values
df['Salary'] = [50000] * len(df)
df
```

```python
## DEleting a column
df.drop('Salary')
## This gives KeyError: "['Salary'] not found in axis" as Pandas by default searches for 'Salary' in rows (axis=0)
```

```python
## To drop a column, we need to specify axis=1
df.drop('Salary', axis=1)
```

```python
df

## But even after dropping the column, we find it still exists in the DataFrame
```

```python
## To permanently delete the column, we need to pass another argument inplace=True
df.drop('Salary', axis=1, inplace=True)
df
```

```python
df
## Now the 'Salary' column has been permanently removed from the DataFrame
## Or we could have assigned the result back to df like this:
## df = df.drop('Salary', axis=1)
```

```python
## Add age to the column
df['Age'] = df['Age'] + 2
df
```

```python
## Delete any value based on row index
df.drop(0)
```

```python
## This again does not delete the row permanently. To do that, we can use inplace=True
df
```

```python
## So to permanently delete the row with index 0, we do:
df.drop(0, axis=0, inplace=True)
df
```

### Handling Missing Values

```python
## Check for null values in the DataFrame
df.isnull()
# Wherever there is a null value, it will show True, else False
```

```python
## To check if any column has null values
df.isnull().any()
```

```python
## To check how many null values are there in each column
df.isnull().sum()
```

```python
## To check which rows have null values
df.isnull().any(axis=1)
```

```python
## We can create a new DataFrame with only those rows which have null values
df_null = df[df.isnull().any(axis=1)]
df_null
```

```python
## To handle null values, we can either drop them or fill them with some value
df_cleaned = df.dropna()  # Drops all rows with any null values
df_cleaned
```

```python
## We can fill null values with a specific value, say 0
df_nulled = df.fillna(0)
df_nulled

## When we replace null values, good practice to create a new DataFrame
```

```python
## We can also fill null values with the mean of that column by creating a column with filled values
df['Sales_filled_mean'] = df['Sales'].fillna(df['Sales'].mean())
df
```

```python
df.dtypes
## To rename a column, we need to give a dictionary with old and new names
df = df.rename(columns={'Date': 'Sale_date'})
df
```

```python
## Change data type of a column
df['Value_new'] = df['Value'].astype(int)
df
## NaN values cannot be converted to int, so those will remain NaN (explained at last)
```

```python
## alternatively fill NaN with mean and then do dtype change using astype()
 
df['Value_new'] = df['Value'].fillna(df['Value'].mean()).astype(int)
df.head()
```

```python
# Apply a function to a column
df['New_value'] = df['Value'].apply(lambda x: x * 2)
df.head()
```

### Data Aggregating and Grouping

```python
df.head()
```

```python
## On the basis of a column, we can group the data
grouped_mean = df.groupby('Product')['Value'].mean()
grouped_mean
```

```python
## We can groupby multiple columns
grouped_multi = df.groupby(['Product', 'Region'])['Value'].sum()
grouped_multi
```

```python
grouped_multi = df.groupby(['Region', 'Product'])['Value'].sum()
grouped_multi
```

```python
## Perform multiple aggregations on grouped data
aggregated = df.groupby('Region')['Value'].agg(['mean', 'sum', 'max', 'count'])
aggregated
```

```python
# Named aggregations

df.groupby('Product').agg(Sales_sum=('Sales','sum'))
```

### MErging and Joining DataFrames

```python
## Creating some DataFrame for practice
df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1, 2, 3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value2': [4, 5, 6]})
```

```python
## Merge df on Key
merged_df = pd.merge(df1, df2, on='Key', how='inner')
merged_df
```

```python
pd.merge(df1, df2, on='Key', how='outer')
```

```python
pd.merge(df1, df2, on='Key', how='left')
```

```python
pd.merge(df1, df2, on='Key', how='right')
```

2
Pandas NaN values are inherently float as it internally stores data as Numpy arrays. Thus Pandas by default converts any numerical column with NaN values into float type.
If we try to convert a NaN value to int, it gives an error as Numpy has no way to represent NaN as int.

New Solution:
Pandas has introduced its own data type called nullable integer types like: Int32, Int64 (capital I) - Pandas dtype. So, use these if cleaning data with missing values and need integers.
This is also true for bool (NumPy) and Bool (Pandas)

General Rule to identify if dtype is of Numpy or Pandas is:\

NumPy dtype starts with small letter and cannot have NaN\
Pandas dtype genrally starts with capital letter and can have NaN
